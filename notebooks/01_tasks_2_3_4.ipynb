```python
from sqlalchemy import create_engine
import pandas as pd
import psycopg2
import boto3
```

Подключение postgres

```python
engine = create_engine("postgresql+psycopg2://admin:admin@postgres:5432/demo_db")
```

```python
wells = pd.read_sql("SELECT * FROM wells", engine)
production  = pd.read_sql("SELECT * FROM production", engine)
telemetry = pd.read_sql("SELECT * FROM well_telemetry", engine)
```

```python
storage_options = {
    "key": "admin",
    "secret": "admin123",
    "client_kwargs": {"endpoint_url": "http://minio:9000"}
}
```

Выгрузка в minio

```python
wells.to_parquet(
    f"s3://raw/wells/",
    partition_cols=["start_date"],
    storage_options=storage_options,
    index=False,
)
production.to_parquet(
    f"s3://raw/production/",
    partition_cols=["date"],
    storage_options=storage_options,
    index=False,
)
telemetry.to_parquet(
    f"s3://raw/telemetry/",
    partition_cols=["timestamp"],
    storage_options=storage_options,
    index=False,
)
```
